```python
df.hist(figsize=(10, 8))
plt.show()
```

```python
sns.pairplot(df, hue='Species')
plt.show()
```

```python
sns.scatterplot(
    x='SepalLengthCm',
    y='SepalWidthCm',
    hue='Species',
    data=df
)
plt.title('Sepal Length vs Sepal Width')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df)
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(
    x='Species',
    y='PetalLengthCm',
    data=df
)
plt.show()
```
