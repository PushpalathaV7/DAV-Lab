```python
import pandas as pd
import numpy as np
```

```python
marks = np.array([85, 25, 35, 72, 68, 95, 70])

print("Original Array")
print(marks)

print("Slice (index 1 to 4):", marks[1:5])
```

```python
print("First mark:", marks[0])
```

```python
print("Last mark:", marks[-1])  
```

```python
print("First three marks:", marks[0:3]) 
```

```python
print("2 to 4 elements:", marks[2:5]) 
```

```python
print("Every second mark:", marks[::2]) 
```

```python
print("Reverse:", marks[::-1]) 
```

```python
d = {
    "Name": ["pushpa", "pragathi", "aishu", "vaishu"],
    "Marks": [92, 74, 85, 63],
    "Branch": ["ISE", "CSE", "AI", "ECE"]
}
data = pd.DataFrame(d)
print(data)
```

```python
print("First 2 Rows\n", data.iloc[:2])
```

```python
print(data.iloc[-1])  
```

```python
print(data.iloc[1:3]) 
```

```python
print("Marks Column:\n", data["Marks"])
```

```python
print("Rows where Marks > 70:\n", data[data["Marks"] > 70])
```
